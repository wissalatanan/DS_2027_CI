### Problématique

1. **Hypothèse 1 (Dynamiques hétérogènes) :** Les exportations des chapitres agricoles (HS 01 à 24) ne suivent pas une évolution linéaire uniforme ; il existe des sous-groupes de produits moteurs à forte croissance et des produits à trajectoire stagnante ou volatile.
2. **Hypothèse 2 (Forme de trajectoire vs Effet de taille) :** La normalisation des séries temporelles (centrage-réduction ou base 100) est nécessaire pour regrouper les produits selon la *forme de leur courbe d'évolution* et non selon leur valeur brute en dollars.
3. **Hypothèse 3 (Complémentarité des modèles) :** L'algorithme des K-Means fournira une partition nette des profils d'exportation, tandis que la classification ascendante hiérarchique (Agglomerative Clustering) permettra de visualiser la structure en arbre et la proximité entre chapitres.

## 3, 4 & 5. Chargement, Dictionnaire des variables et Nettoyage

### Dictionnaire des variables
- `annee` (int) : Année de l'observation (2010–2024)[cite: 2].
- `code_hs` (str) : Code du chapitre du Système Harmonisé (HS 01 à HS 24)[cite: 1].
- `libelle_hs` (str) : Description de la catégorie de produits agroalimentaires.
- `valeur_usd` (float) : Valeur totale des exportations en dollars US.
- `valeur_norm` (float) : Valeur normalisée en base 100 (Année 2010 = 100) pour l'analyse de trajectoire[cite: 1].

```python
# ÉTAPE 3, 4 & 5 : CRÉATION DU DATAFRAME ET NETTOYAGE
# Génération structurée des données 2010-2024 pour HS01 à HS24
np.random.seed(42)

libelles_hs = {
    'HS01': 'Animaux vivants', 'HS02': 'Viandes', 'HS03': 'Poissons et crustacés',
    'HS04': 'Produits laitiers', 'HS05': 'Autres produits d origine animale',
    'HS06': 'Plantes et fleurs', 'HS07': 'Légumes et racines', 'HS08': 'Fruits comestibles',
    'HS09': 'Café, thé, épices', 'HS10': 'Céréales', 'HS11': 'Produits de la meunerie',
    'HS12': 'Graines oléagineuses', 'HS13': 'Gommes et résines', 'HS14': 'Matières à tresser',
    'HS15': 'Graisses et huiles', 'HS16': 'Préparations de viandes/poissons',
    'HS17': 'Sucres et sucreries', 'HS18': 'Cacao et préparations',
    'HS19': 'Préparations à base de céréales', 'HS20': 'Préparations de légumes/fruits',
    'HS21': 'Préparations alimentaires diverses', 'HS22': 'Boissons et vinaigres',
    'HS23': 'Résidus des industries alimentaires', 'HS24': 'Tabacs'
}

records = []
annees = list(range(2010, 2025))

for code, lib in libelles_hs.items():
    # Base aléatoire réaliste selon le secteur
    val_base = np.random.uniform(10, 500) * 1e6
    tendance = np.random.uniform(-0.02, 0.08)

    for i, annee in enumerate(annees):
        bruit = np.random.normal(0, 0.05)
        valeur = val_base * ((1 + tendance) ** i) * (1 + bruit)
        records.append({
            'annee': annee,
            'code_hs': code,
            'libelle_hs': lib,
            'valeur_usd': max(valeur, 1e5) # Éviter les valeurs négatives
        })

df = pd.DataFrame(records)

# Traitement des valeurs manquantes et doublons
assert df.isnull().sum().sum() == 0, "Présence de valeurs manquantes !"
df = df.drop_duplicates()

# Construction de la matrice Pivot (Chapitres en lignes, Années en colonnes)
df_pivot = df.pivot(index='code_hs', columns='annee', values='valeur_usd')

# Normalisation en Base 100 (Année 2010 = 100) pour comparer la forme des séries
df_norm = df_pivot.div(df_pivot[2010], axis=0) * 100

print(f" Dimensions de la matrice des séries : {df_pivot.shape} (24 chapitres x 15 années)")
df_norm.head()
```

## 7, 8, 9, 10 & 11. Modélisation, Validation et Comparaison

- **Séparation temporelle :**
  - Train (2010–2019) : Entraînement et création des clusters.
  - Test / Validation (2020–2024) : Test de stabilité des profils.
- **Modèle de référence :** K-Means naïf ($K=2$).
- **Modèle alternatif 1 :** K-Means optimisé ($K$ optimal par le score de Silhouette).
- **Modèle alternatif 2 :** Agglomerative Clustering (Classification Hiérarchique).

```python
# ÉTAPE 7 À 12 : CLUSTERING, SILHOUETTE ET COMPARISON
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score
from scipy.cluster.hierarchy import dendrogram, linkage

# 1. Séparation temporelle
X_train = df_norm.loc[:, 2010:2019]
X_test = df_norm.loc[:, 2020:2024]
X_full = df_norm.copy()

# 2. Recherche du nombre optimal de clusters K (Méthode de la Silhouette)
silhouettes_km = []
K_range = range(2, 7)

for k in K_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_train)
    silhouettes_km.append(silhouette_score(X_train, labels))

best_k = K_range[np.argmax(silhouettes_km)]
print(f" Nombre optimal de clusters (Silhouette K-Means) : K = {best_k}")

# 3. Entraînement des Modèles
# Modèle Baseline (K-Means K=2)
baseline = KMeans(n_clusters=2, random_state=42, n_init=10)
labels_base = baseline.fit_predict(X_full)
score_base = silhouette_score(X_full, labels_base)

# Modèle 1 : K-Means Optimisé
km_opt = KMeans(n_clusters=best_k, random_state=42, n_init=10)
labels_km = km_opt.fit_predict(X_full)
score_km = silhouette_score(X_full, labels_km)

# Modèle 2 : Agglomerative Clustering (Hiérarchique)
agg = AgglomerativeClustering(n_clusters=best_k, metric='euclidean', linkage='ward')
labels_agg = agg.fit_predict(X_full)
score_agg = silhouette_score(X_full, labels_agg)

# 4. Tableau comparatif des performances
comparaison = pd.DataFrame([
    {'Modele': 'Référence (K-Means K=2)', 'Clusters (K)': 2, 'Score Silhouette': score_base},
    {'Modele': 'K-Means Optimisé', 'Clusters (K)': best_k, 'Score Silhouette': score_km},
    {'Modele': 'Agglomerative Clustering', 'Clusters (K)': best_k, 'Score Silhouette': score_agg}
])

display(comparaison)
```
